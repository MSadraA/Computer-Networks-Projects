## Loading results

`load_vectors(config)` returns every vector of one config, one row per run and vector, with `repetition` as a column.
`vector(vecs, rep, module, name)` returns the times and values of one vector in run `rep` (two empty arrays if that run did not record it).

Vectors you will need (full module paths start with `Roaming.`):

| config | module | vector | one sample per |
|---|---|---|---|
| `Survey` | `meter.wlan[0].radio`, `meter.wlan[1].radio` | `minSnir:vector` | beacon of ap1 / ap2 the meter received (SNIR as a ratio, not dB) |
| `Walk` | `ap1.sink`, `ap2.sink` | `packets:vector(packetLength)` | voice frame delivered by that AP |
| `Walk` | `host.wlan.agent` | `sentRequest:vector`, `acceptConfirm:vector`, `dropConfirm:vector` | request to / answer from the management module (the value is a code) |
| `Walk` | `host.wlan.radio` | `radioChannel:vector` | channel change of the station's radio |
| `Walk` | `host.wlan.radio` | `minSnir:vector` | frame the station received |
| `Walk` | `host.wlan.mac.dcf` | `packetSentToPeerWithoutRetry:vector(packetBytes)`, `packetSentToPeerWithRetry:vector(packetBytes)` | first transmission / retransmission of a frame |

```python
RESULTS = "results"


def load_vectors(config):
    df = results.read_result_files(os.path.join(RESULTS, f"{config}-*.vec"))
    vecs = results.get_vectors(df, include_itervars=True, include_runattrs=True)
    vecs["repetition"] = pd.to_numeric(vecs["repetition"])
    return vecs


def vector(vecs, rep, module, name):
    row = vecs[(vecs.repetition == rep) & (vecs.module == "Roaming." + module) & (vecs.name == name)]
    if row.empty:
        return np.array([]), np.array([])
    return row.iloc[0].vectime, row.iloc[0].vecvalue
```

## 3-1  Anatomy of a handover

```python

import pandas as pd

# Load Walk simulation results
walk = load_vectors("Walk")

# Select the first repetition
rep = 0

# --------------------------------------------------
# 1. Extract required vectors
# --------------------------------------------------

# Packet reception times at both APs
t_ap1, _ = vector(
    walk, rep, "ap1.sink", "packets:vector(packetLength)"
)

t_ap2, _ = vector(
    walk, rep, "ap2.sink", "packets:vector(packetLength)"
)

# Management requests and confirmations
t_req, v_req = vector(
    walk, rep, "host.wlan.agent", "sentRequest:vector"
)

t_conf, v_conf = vector(
    walk, rep, "host.wlan.agent", "acceptConfirm:vector"
)

# Radio channel changes
t_chan, v_chan = vector(
    walk, rep, "host.wlan.radio", "radioChannel:vector"
)

# --------------------------------------------------
# 2. Validate AP reception data
# --------------------------------------------------

if len(t_ap1) == 0:
    raise ValueError("No frames were received by ap1 in this run.")

if len(t_ap2) == 0:
    raise ValueError("No frames were received by ap2 in this run.")

# Last frame received by ap1 and first frame received by ap2
last_ap1_time = t_ap1[-1]
first_ap2_time = t_ap2[0]

if last_ap1_time >= first_ap2_time:
    raise ValueError(
        "The AP reception times do not define a valid handover interval."
    )

# --------------------------------------------------
# 3. Build the handover timeline
# --------------------------------------------------

timeline = []

# Last frame received by ap1
timeline.append({
    "Time (s)": last_ap1_time,
    "Event": "Last voice frame received by ap1"
})

# Management requests:
# 1 = Scan, 2 = Authentication, 4 = Association
for t, v in zip(t_req, v_req):

    if not (last_ap1_time < t < first_ap2_time):
        continue

    if v == 1:
        event = "Scan request (start of scanning)"

    elif v == 2:
        event = "End of scanning / Authentication request"

    elif v == 4:
        event = "Association request"

    else:
        continue

    timeline.append({
        "Time (s)": t,
        "Event": event
    })

# Radio channel changes during scanning
for t, v in zip(t_chan, v_chan):

    if last_ap1_time < t < first_ap2_time:
        timeline.append({
            "Time (s)": t,
            "Event": f"Start listening on channel {v}"
        })

# Management confirmations:
# 2 = Authentication confirmation
# 4 = Association confirmation
for t, v in zip(t_conf, v_conf):

    if not (last_ap1_time < t < first_ap2_time):
        continue

    if v == 2:
        event = "Authentication confirmed"

    elif v == 4:
        event = "Association confirmed"

    else:
        continue

    timeline.append({
        "Time (s)": t,
        "Event": event
    })

# First frame received by ap2
timeline.append({
    "Time (s)": first_ap2_time,
    "Event": "First voice frame received by ap2"
})

# --------------------------------------------------
# 4. Sort and display the timeline
# --------------------------------------------------

df_timeline = (
    pd.DataFrame(timeline)
    .sort_values(by="Time (s)", kind="stable")
    .reset_index(drop=True)
)

pd.set_option("display.max_rows", None)
pd.set_option("display.max_colwidth", None)

display(df_timeline)
```

```python

import numpy as np
import scipy.stats as st

walk_vecs = load_vectors("Walk")
outage_times = []

for rep in range(5):
    t_ap1, _ = vector(
        walk_vecs, rep, "ap1.sink",
        "packets:vector(packetLength)"
    )

    t_ap2, _ = vector(
        walk_vecs, rep, "ap2.sink",
        "packets:vector(packetLength)"
    )

    times_ap1 = np.asarray(t_ap1, dtype=float)
    times_ap2 = np.asarray(t_ap2, dtype=float)

    if len(times_ap1) == 0 or len(times_ap2) == 0:
        raise ValueError(f"Missing packet data in repetition {rep + 1}")

    outage = np.min(times_ap2) - np.max(times_ap1)

    if outage < 0:
        raise ValueError(f"Invalid outage in repetition {rep + 1}")

    outage_times.append(outage)

mean_outage = np.mean(outage_times)
std_outage = np.std(outage_times, ddof=1)
sem_outage = st.sem(outage_times)

ci_95 = st.t.interval(
    confidence=0.95,
    df=len(outage_times) - 1,
    loc=mean_outage,
    scale=sem_outage
)

print("Outage times for 5 runs:")
for i, outage in enumerate(outage_times, start=1):
    print(f"Run {i}: {outage:.6f} s")

print(f"\nMean outage time: {mean_outage:.6f} s")
print(f"Standard deviation: {std_outage:.6f} s")
print(f"95% Confidence Interval: ({ci_95[0]:.6f}, {ci_95[1]:.6f}) s")

nch = 7
probeDelay = 0.100
minChannelTime = 0.150
maxChannelTime = 0.300

n_active = 2
n_inactive = nch - n_active

scan_time = (
    nch * probeDelay
    + n_inactive * minChannelTime
    + n_active * maxChannelTime
)

print(f"\nAnalytical scanning time: {scan_time:.6f} s")
print(f"Mean measured outage time: {mean_outage:.6f} s")
print(f"Difference: {mean_outage - scan_time:.6f} s")
```

```python
import numpy as np

neighbor_vecs = load_vectors("ScanNeighbor")
outage_times_neighbor = []

for rep in range(5):
    t_ap1, _ = vector(neighbor_vecs, rep, "ap1.sink", "packets:vector(packetLength)")
    t_ap2, _ = vector(neighbor_vecs, rep, "ap2.sink", "packets:vector(packetLength)")
    
    if len(t_ap1) > 0 and len(t_ap2) > 0:
        last_ap1 = t_ap1[-1] 
        first_ap2 = t_ap2[0]
        outage = first_ap2 - last_ap1
        outage_times_neighbor.append(outage)

mean_outage_neighbor = np.mean(outage_times_neighbor)

print("Outage times for ScanNeighbor (seconds):", [round(o, 4) for o in outage_times_neighbor])
print(f"Mean Outage Time (ScanNeighbor): {mean_outage_neighbor:.4f} seconds")
print(f"Difference from prediction (0.8s): {mean_outage_neighbor - 0.8:.4f} seconds")
```

## 3-2  The sticky client

```python
import numpy as np
import matplotlib.pyplot as plt

D = 51.0
v = 1.75
alpha = 3.2
sigma = 3.0
Pt_dBm = 10 * np.log10(20) # 20mW = 13.01 dBm
noise_dBm = -110.0 

PL_d0 = 46.4

survey_vecs = load_vectors("Survey")
rep = 0 

t_ap1, snir_ap1 = vector(survey_vecs, rep, "meter.wlan[0].radio", "minSnir:vector")
t_ap2, snir_ap2 = vector(survey_vecs, rep, "meter.wlan[1].radio", "minSnir:vector")

t_walk = D / v
valid_ap1 = t_ap1 <= t_walk
valid_ap2 = t_ap2 <= t_walk

t1, s1 = t_ap1[valid_ap1], snir_ap1[valid_ap1]
t2, s2 = t_ap2[valid_ap2], snir_ap2[valid_ap2]

pr_ap1 = 10 * np.log10(s1) + noise_dBm
pr_ap2 = 10 * np.log10(s2) + noise_dBm

x1 = v * t1
d1 = np.sqrt(x1**2 + 2**2)

x2 = v * t2
d2 = np.sqrt((D - x2)**2 + 2**2)

plt.figure(figsize=(10, 6))

plt.scatter(d1, pr_ap1, alpha=0.5, label='Measured ap1', s=15, color='blue')
plt.scatter(d2, pr_ap2, alpha=0.5, label='Measured ap2', s=15, color='green')

d_range = np.logspace(np.log10(2), np.log10(np.sqrt(D**2 + 2**2)), 100)
pr_mean = Pt_dBm - PL_d0 - 10 * alpha * np.log10(d_range)
plt.plot(d_range, pr_mean, color='red', linewidth=3, label='Mean Model $\\bar{P}(d)$')

plt.axhline(-85, color='black', linestyle='--', linewidth=2, label='Sensitivity (-85 dBm)')

plt.xscale('log') 
plt.xlabel('Distance (m) - Log Scale')
plt.ylabel('Received Power (dBm)')
plt.title('Received Power vs Distance with Log-Normal Shadowing')
plt.legend()
plt.grid(True, which="both", ls="--")
plt.show()
```

```python
import numpy as np

D = 51.0
v = 1.75
ideal_x = D / 2.0
ideal_t = ideal_x / v

walk_vecs = load_vectors("Walk")
scan_times = []
scan_distances = []
delay_meters = []
delay_seconds = []

for rep in range(5):
    t_req, v_req = vector(walk_vecs, rep, "host.wlan.agent", "sentRequest:vector")
    
    scan_time = None
    for t, val in zip(t_req, v_req):
        if val == 1 and t > 1.0:
            scan_time = t
            break
            
    scan_times.append(scan_time)
    
    scan_x = scan_time * v
    scan_distances.append(scan_x)
    
    delay_m = scan_x - ideal_x
    delay_s = scan_time - ideal_t
    
    delay_meters.append(delay_m)
    delay_seconds.append(delay_s)

print(f"Ideal Switch Point: {ideal_x} m (at {ideal_t:.2f} s)")
print("-" * 40)
for i in range(5):
    print(f"Run {i+1}: Scan started at x = {scan_distances[i]:.2f} m (t = {scan_times[i]:.4f} s)")
    print(f"         Sticky Delay = {delay_meters[i]:.2f} m ({delay_seconds[i]:.2f} s)")

print("-" * 40)
print(f"Average Sticky Delay: {np.mean(delay_meters):.2f} m ({np.mean(delay_seconds):.2f} s)")
```

## 3-3  Your roaming algorithm

```python

import numpy as np
import matplotlib.pyplot as plt

noise_dBm = -110.0
v = 1.75
D = 51.0
BI = 0.1
timeout = 3.5 * BI

T_thresh = -75.0
H_values = [0, 2, 4, 6, 8, 10]
N_values = [1, 10]

survey_vecs = load_vectors("Survey")


def run_state_machine(rep, H, N):
    t1, snir1 = vector(
        survey_vecs, rep, "meter.wlan[0].radio", "minSnir:vector"
    )
    t2, snir2 = vector(
        survey_vecs, rep, "meter.wlan[1].radio", "minSnir:vector"
    )

    t1 = np.asarray(t1, dtype=float)
    t2 = np.asarray(t2, dtype=float)
    snir1 = np.asarray(snir1, dtype=float)
    snir2 = np.asarray(snir2, dtype=float)

    p1 = 10 * np.log10(snir1) + noise_dBm
    p2 = 10 * np.log10(snir2) + noise_dBm

    valid1 = np.isfinite(t1) & np.isfinite(p1) & (snir1 > 0)
    valid2 = np.isfinite(t2) & np.isfinite(p2) & (snir2 > 0)

    events = (
        [(t, 1, p) for t, p in zip(t1[valid1], p1[valid1])] +
        [(t, 2, p) for t, p in zip(t2[valid2], p2[valid2])]
    )
    events.sort(key=lambda e: (e[0], e[1]))

    buffers = {1: [], 2: []}
    latest_frame = {1: None, 2: None}

    current_ap = 1
    timer_ref = None

    handovers = 0
    last_handover_time = None

    for t, ap_id, power in events:

        # Check whether the current AP has timed out
        while (
            current_ap is not None
            and timer_ref is not None
            and t - timer_ref >= timeout
        ):
            timeout_time = timer_ref + timeout

            candidates = [
                ap for ap in (1, 2)
                if latest_frame[ap] is not None
            ]

            if not candidates:
                current_ap = None
                timer_ref = None
                break

            selected_ap = max(
                candidates,
                key=lambda ap: latest_frame[ap][1]
            )

            if selected_ap != current_ap:
                handovers += 1
                last_handover_time = timeout_time

            current_ap = selected_ap
            timer_ref = timeout_time

        # Record the newly received beacon
        latest_frame[ap_id] = (t, power)

        buffers[ap_id].append(power)
        if len(buffers[ap_id]) > N:
            buffers[ap_id].pop(0)

        # If scanning, associate with the first available AP
        if current_ap is None:
            current_ap = ap_id
            timer_ref = t

        # A beacon from the current AP resets its timeout
        elif ap_id == current_ap:
            timer_ref = t

        # Evaluate handover only after both SMA buffers are full
        if len(buffers[1]) == N and len(buffers[2]) == N:
            P_est1 = np.mean(buffers[1])
            P_est2 = np.mean(buffers[2])

            if current_ap == 1:
                if P_est1 < T_thresh and P_est2 >= P_est1 + H:
                    current_ap = 2
                    handovers += 1
                    last_handover_time = t
                    timer_ref = t

            elif current_ap == 2:
                if P_est2 < T_thresh and P_est1 >= P_est2 + H:
                    current_ap = 1
                    handovers += 1
                    last_handover_time = t
                    timer_ref = t

    if last_handover_time is None:
        last_handover_x = None
    else:
        last_handover_x = min(last_handover_time * v, D)

    return handovers, last_handover_x


results = {
    N: {"H": [], "avg_ho": [], "avg_loc": []}
    for N in N_values
}

print(f"Threshold T = {T_thresh} dBm")
print(f"Beacon timeout = {timeout:.3f} s")

for N in N_values:
    print(f"\nResults for N = {N}:")

    for H in H_values:
        ho_list = []
        loc_list = []

        for rep in range(3):
            ho, loc = run_state_machine(rep, H, N)

            ho_list.append(ho)

            if loc is not None:
                loc_list.append(loc)

        avg_ho = np.mean(ho_list)
        avg_loc = np.mean(loc_list) if loc_list else np.nan

        results[N]["H"].append(H)
        results[N]["avg_ho"].append(avg_ho)
        results[N]["avg_loc"].append(avg_loc)

        loc_text = f"{avg_loc:.2f} m" if np.isfinite(avg_loc) else "N/A"

        print(
            f"H = {H:2d} dB | "
            f"Average handovers = {avg_ho:.2f} | "
            f"Average last handover location = {loc_text}"
        )


plt.figure(figsize=(8, 5))

plt.plot(
    results[1]["H"],
    results[1]["avg_ho"],
    marker="o",
    label="N = 1 (No Smoothing)"
)

plt.plot(
    results[10]["H"],
    results[10]["avg_ho"],
    marker="s",
    label="N = 10 (Smoothed)"
)

plt.xlabel("Hysteresis H (dB)")
plt.ylabel("Average Number of Handovers")
plt.title(f"Handovers vs Hysteresis (T = {T_thresh} dBm)")
plt.xticks(H_values)
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
