## 0. Load and access simulation results

The setup cell reads the group parameters. The helper functions in the next code cell provide access to the scalar and vector result files stored in `results/`.

- `load_scalars(config)` loads scalar results for a configuration and includes iteration variables and run attributes.
- `scalar(sc, module, name)` selects a scalar by module-path pattern and name, then sums matching values per run.
- `load_vector(config, module, name)` retrieves the time and value arrays for a selected vector.


```python
RESULTS = "results"


def load_scalars(config):
    df = results.read_result_files(os.path.join(RESULTS, f"{config}-*.sca"))
    sc = results.get_scalars(df, include_itervars=True, include_runattrs=True)
    for col in ("n", "repetition"):
        if col in sc:
            sc[col] = pd.to_numeric(sc[col])
    return sc


def scalar(sc, module, name):
    rows = sc[(sc.name == name) & sc.module.str.fullmatch(module)]
    keys = [c for c in ("n", "rtsThreshold", "repetition") if c in rows and rows[c].notna().all()]
    return rows.groupby(keys).value.sum()


def load_vector(config, module, name):
    df = results.read_result_files(os.path.join(RESULTS, f"{config}-*.vec"))
    v = results.get_vectors(df)
    row = v[(v.module == module) & (v.name == name)].iloc[0]
    return row.vectime, row.vecvalue
```

### Quick check — Hidden configuration

Example lookup: count the frames received by the AP in each run of the `Hidden` configuration. Run `Hidden` in OMNeT++ before executing the example cell.


```python
hidden = load_scalars("Hidden")
scalar(hidden, r"Contention\.ap\.sink", "packets:count")
```
